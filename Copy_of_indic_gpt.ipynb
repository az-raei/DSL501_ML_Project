```python
import json
import random
from datetime import datetime
from transformers import pipeline

pipe = pipeline("text-generation", model="aashay96/indic-gpt", tokenizer="aashay96/indic-gpt")

topics = [
    "ఒంటరితనం", "ఆందోళన", "దుఃఖం", "ప్రేమ విరహం",
    "పని ఒత్తిడి", "కుటుంబ సమస్యలు", "స్వీయ విశ్వాసం", "ఆరోగ్యం", "మానసిక శాంతి"
]
```

```python


def generate_conversation(topic):
    """Generate a short empathetic Telugu conversation."""
    seeker_prompt = f"విషయం: {topic}\nరోగి (సమస్య చెబుతున్నాడు):"
    response_prompt = f"విషయం: {topic}\nథెరపిస్టు (సానుభూతి గల సమాధానం ఇస్తున్నాడు):"

    # Generate seeker post
    seeker_text = pipe(
        seeker_prompt,
        max_new_tokens=60,
        temperature=0.8,
        top_p=0.95,
        do_sample=True
    )[0]["generated_text"]

    # Generate therapist response
    response_text = pipe(
        response_prompt,
        max_new_tokens=80,
        temperature=0.8,
        top_p=0.95,
        do_sample=True
    )[0]["generated_text"]

    # Clean generated text
    seeker_post = seeker_text.replace(seeker_prompt, "").strip()
    response = response_text.replace(response_prompt, "").strip()

    # Empathy levels (dummy random values for now)
    empathy_levels = {
        "emotional_reaction": random.randint(0, 2),
        "interpretation": random.randint(0, 2),
        "exploration": random.randint(0, 2)
    }

    return {
        "topic": topic,
        "seeker_post": seeker_post,
        "response": response,
        "empathy_levels": empathy_levels
    }
```

```python
dataset = []
save_every = 10
file_index = 1

for i in range(50):
    topic = random.choice(topics)
    data_instance = generate_conversation(topic)
    dataset.append(data_instance)

    if len(dataset) % save_every == 0:
        filename = f"telugu_empathy_dataset_{file_index}_{datetime.now().strftime('%Y%m%d_%H%M%S')}.json"
        with open(filename, "w", encoding="utf-8") as f:
            json.dump(dataset, f, ensure_ascii=False, indent=2)
        print(f"✅ Saved {len(dataset)} conversations to {filename}")
        dataset = []
        file_index += 1
```
